```python
import math
import matplotlib.pyplot as plt
import numpy as np

m = 15.59
v0 = 129
h0 = 250
t0 = 3

CdA = 0.34 * math.pi* (0.124/2)**2
g = 9.81

import numpy as np

def ascent(v0, g, rho, CdA, m):
    k = rho * CdA / (2*m)
    w = np.sqrt(g*k)
    th0 = np.arctan(v0*np.sqrt(k/g))
    t_apo = th0 / w
    h_apo = np.log1p(k*v0**2/g) / (2*k)
    h = lambda t: np.log(np.cos(th0 - w*t) / np.cos(th0)) / k
    v = lambda t: np.sqrt(g/k) * np.tan(th0 - w*t)
    return t_apo, h_apo, h, v

t_apo, h_apo, h, v = ascent(v0,g,1,CdA,m) 
print(f"Apogee of {(h_apo+h0):.2f}m at {(t_apo+t0):.2f}s")
```
