### Process noise
To get a good estimate on the expected pn, we should have a rough estimate of the jerk the rocket will experience (max) which will be right at the start of the simulation so at motor burnout. We will numerically solve the ODE that describes the drag that changes and varies with it the acceleration.

$$
\dot v = -9.81 - \frac{1}{2m}C_dApv²
$$

```python
import math
from scipy import integrate
import matplotlib.pyplot as plt

p = 1.225
A = math.pi*0.06**2
C_d = 0.35
m = 20

v0 = 130

def fun(t,y):
    v = y[0]


    dv_dt = -9.81 - (0.5/m) * C_d * A * p * v**2
    a = dv_dt
    return [dv_dt]


res = integrate.solve_ivp(fun, (0,5), [v0])

plt.plot(res.t, res.y[0,:])
plt.show()
```
